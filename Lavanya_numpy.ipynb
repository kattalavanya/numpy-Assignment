A1. Why are NumPy arrays preferred over Python lists?


NumPy arrays are preferred over Python lists because they are faster and more memory-efficient for numerical computations. They also support vectorized operations, which allow mathematical calculations on entire arrays without explicit Python loops. Therefore, NumPy is more suitable for handling large numerical datasets in AI/ML.


A2. Explain vectorization. Why does replacing an explicit Python loop with a vectorized NumPy operation matter with large datasets? 


Vectorization means performing an operation on an entire NumPy array at once instead of using an explicit Python loop. It is faster because NumPy uses optimized operations to process many elements efficiently. For large datasets, vectorization reduces execution time and makes numerical computations more efficient.

A3. For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each


A * B performs element-wise multiplication, where corresponding elements of A and B are multiplied. A @ B performs matrix multiplication, following matrix multiplication rules. In a neural network, A * B can be used for element-wise operations such as applying a mask, while A @ B is commonly used to calculate the weighted sum of inputs in a layer.

A4. What is broadcasting in NumPy? State the general rule used to decide whether arrays of different shapes can be combined element-wise

Broadcasting allows NumPy to perform element-wise operations on arrays with different but compatible shapes. Two dimensions are compatible when they are equal or when one of them is 1. NumPy automatically expands the smaller dimension conceptually to perform the operation.

A5. Explain flatten() vs ravel(). In a memory-constrained application processing large images, which would you prefer and why? 

Both flatten() and ravel() convert an array into a one-dimensional array. flatten() always creates a copy, while ravel() generally returns a view when possible. For memory-constrained processing of large images, I would prefer ravel() because it can avoid creating an unnecessary copy and use less memory.

A7. Why is it important to normalize or standardize numerical features before training most ML models? Explain normalization vs standardization.

Normalization or standardization is important because numerical features may have very different scales. For example, age may range from 18 to 60, while salary may range from 20,000 to 100,000. Scaling the features puts them on comparable scales and can help many ML algorithms work more effectively.

Normalization scales values to a fixed range, usually 0 to 1, using (x - min) / (max - min). Standardization transforms values so that they have a mean of approximately 0 and a standard deviation of 1, using (x - mean) / standard deviation.

If np.mean() is directly applied to an array containing NaN, the result will generally be NaN.

```python
#C1. Pass/Fail using np.where()
import numpy as np

scores = np.array([35, 45, 67, 28, 90])

labels = np.where(scores >= 40, 'P', 'F')

print(labels)
```

```python
#C2. Standardization function
import numpy as np

def standardize(x):
    mean = np.mean(x)
    std = np.std(x)
    return (x - mean) / std

x = np.array([10, 20, 30, 40, 50])

result = standardize(x)

print(result)
```

Explanation: Here, I calculated the mean and standard deviation of the values and used them to standardize the data. After standardization, the values are represented based on how far they are from the mean. The middle value becomes 0, while values below and above the mean get negative and positive values.

```python
#C3. Combine two batches
import numpy as np

batch1 = np.array([[1, 2, 3],
                   [4, 5, 6]])

batch2 = np.array([[7, 8, 9],
                   [10, 11, 12]])

combined = np.vstack((batch1, batch2))

print(combined)
print(combined.shape)
```

Explanation: Here, both batches have 3 columns, so I used vstack() to place the second batch below the first batch. The two 2×3 arrays become one 4×3 array. I understood that vstack() combines arrays vertically, meaning row by row.

```python
#C4. Standardize X with shape (100, 5)
import numpy as np

X = np.random.default_rng(42).random((100, 5))

mean = X.mean(axis=0)
std = X.std(axis=0)

X_standardized = (X - mean) / std

print(X_standardized.shape)
```

Explanation: Here, I created a dataset with 100 rows and 5 features. Using axis=0, I calculated the mean and standard deviation separately for each column and then standardized the values. The shape remains (100,5) because standardization changes the values, not the number of rows or columns.

```python
#C5. Linear prediction and MSE
import numpy as np

X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])
b = 2.0

y_true = np.array([4, 6, 8])

y_pred = X @ w + b

mse = np.mean((y_true - y_pred) ** 2)

print("Predictions:", y_pred)
print("MSE:", mse)
```

Explanation: Here, I used matrix multiplication between X and w and then added the bias b to get the predicted values. I then compared these predictions with the actual values using Mean Squared Error (MSE). The MSE is small because the predicted values are close to the actual values.

D1. Data Preprocessing Pipeline

You are given a raw 1D or 2D NumPy array containing NaN values.
Describe, step by step, the NumPy operations you would apply to:
(a) detect missing values,
(b) handle them appropriately and justify your strategy, and
(c) normalize or standardize the clean data before it is fed to a model.

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.


Steps

Use np.isnan() to detect missing values.
Calculate a suitable replacement value, such as the mean of the non-missing values.
Replace the missing values using np.where().
Standardize the cleaned data using its mean and standard deviation.

```python
import numpy as np
data = np.array([10.0, 20.0, np.nan, 40.0, 50.0])

# Step 1: Detect missing values
missing = np.isnan(data)
print("Missing values:", missing)

# Step 2: Calculate mean excluding NaN 
mean_value = np.nanmean(data)

# Step 3: Replace NaN with the mean 
clean_data = np.where(missing, mean_value, data)

# Step 4: Standardize the clean data
mean = np.mean(clean_data)
std = np.std(clean_data)
standardized_data = (clean_data - mean) / std

print("Clean data:", clean_data)
print("Standardized data:", standardized_data)
```

Explanation: Here, I created a batch of 16 RGB images, where each image has a size of 64 × 64. Using mean(axis=0), I calculated the average value of all 16 images pixel by pixel. Therefore, the batch dimension is removed and the average image has the shape (64,64,3).

                 D3 — Debug the Model Code

Bug 1 — Weight vector has the wrong shape

X has shape:

(3, 2)

Therefore, the weight vector must contain 2 weights, one for each feature.

However:

w = np.array([0.5, 1.0, 0.2])

has shape:

(3,)

Matrix multiplication:

(3, 2) @ (3,)

is not compatible.

Therefore, w should have shape (2,).

Bug 2 — Incorrect MSE formula

The original expression is:

np.mean(y_true - y_pred ** 2)

Because exponentiation has higher precedence, Python interprets this as:

np.mean(y_true - (y_pred ** 2))

But MSE requires the difference to be squared:

np.mean((y_true - y_pred) ** 2)

```python
import numpy as np

X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

# Two features -> two weights
w = np.array([0.5, 1.0])

b = 2.0

# Prediction
y_pred = X @ w + b

y_true = np.array([10, 20, 30])

# Correct MSE formula
mse = np.mean((y_true - y_pred) ** 2)

print("Predictions:", y_pred)
print("MSE:", mse)

```

Explanation:Here, I understood that the number of weights must match the number of columns in X. Since X has 2 columns, w should contain 2 values. I also corrected the MSE formula by putting the complete difference inside brackets before squaring, so the error is calculated correctly.

```python
#E1. Per-Column Min-Max Normalization
import numpy as np

X = np.array([[10, 100, 1000],
              [20, 200, 2000],
              [30, 300, 3000]])

column_min = X.min(axis=0)
column_max = X.max(axis=0)

normalized = (X - column_min) / (column_max - column_min)

print(normalized)
```

Explanation: Here, I performed min-max normalization separately for each column by using axis=0. The smallest value in each column becomes 0 and the largest becomes 1. This makes the different features comparable by putting them into the same range.
